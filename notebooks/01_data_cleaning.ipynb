## Data Cleaning


```python
import numpy as np 
import pandas as pd 
```

```python
# Reading Files
cpi = pd.read_excel("../data/raw/CPI.xlsx", header=2)
wpi = pd.read_excel("../data/raw/WPI.xlsx")
```

### CPI Cleaning

```python
cpi.columns = ['Date', 'CPI', 'Food', 'Food_Bev', 'Tobacco', 'Clothing', 'Housing', 'Fuel', 'Misc']
```

```python
cpi = cpi.iloc[1:]
```

```python
cpi.head()
```

```python
cpi = cpi.reset_index(drop=True)
```

```python
cpi.head()
```

```python
cpi['Date'] = pd.to_datetime(cpi['Date'])
cpi['Date'] = cpi['Date'].dt.date
```

```python
cpi = cpi.iloc[:, :5]
```

```python
cpi.head()
```

```python
# Convert types
cpi['Date'] = pd.to_datetime(cpi['Date'])

cpi['CPI'] = pd.to_numeric(cpi['CPI'])
cpi['Food'] = pd.to_numeric(cpi['Food'])
cpi['Food_Bev'] = pd.to_numeric(cpi['Food_Bev'])
cpi['Tobacco'] = pd.to_numeric(cpi['Tobacco'])
```

```python
cpi.info()
```

```python
cpi = cpi[['Date', 'CPI']]
```

```python
cpi.head()
```

### WPI Cleaning

```python
wpi.head()
```

```python
wpi = pd.read_excel("../data/raw/WPI.xlsx", header=3)
```

```python
wpi.head()
```

```python
wpi.columns
```

```python
wpi = pd.read_excel("../data/raw/WPI.xlsx", skiprows=3)
```

```python
wpi.head()
```

```python
wpi = wpi[wpi.iloc[:,0] != 'Weight=>']
```

```python
wpi.head()
```

```python
wpi = wpi.iloc[:, :2]   # safest
```

```python
wpi.head()
```

```python
wpi.columns = ['Date', 'WPI']
```

```python
wpi = wpi.iloc[2:]
```

```python
wpi = wpi.reset_index(drop=True)
```

```python
wpi.head()
```

```python
wpi.info()
```

```python
wpi.count()
```

```python
wpi['Date'] = pd.to_datetime(wpi['Date'], format='%b-%Y', errors='coerce')
wpi['WPI'] = pd.to_numeric(wpi['WPI'], errors='coerce')
```

```python
wpi['WPI'] = wpi['WPI'].ffill()
```

```python
wpi.head()
```

```python
wpi['WPI'] = wpi['WPI'].ffill().bfill()
```

```python
wpi = wpi[wpi['Date'] >= '2012-04-01']
```

```python
wpi = wpi.reset_index(drop=True)
```

```python
cpi.head()
wpi.head()
```

```python
cpi.head()
```

```python
cpi = cpi[cpi['Date'] >= '2012-04-01']
cpi = cpi.reset_index(drop=True)
```

```python
df = pd.merge(cpi, wpi, on='Date', how='inner')
```

```python
df.to_csv("../data/processed/final_dataset.csv", index=False)
```
